```python
import requests
from Bio.PDB import DSSP, PDBParser  
import pandas as pd
import numpy as np
import requests
import json
import os
from tqdm import tqdm
import time
import warnings  
from urllib3.exceptions import InsecureRequestWarning 
warnings.simplefilter("ignore", category=FutureWarning)  
warnings.simplefilter("ignore", category=InsecureRequestWarning)  
```

```python
class esmfold_request:  
    def __init__(self, path):  
        if not os.path.exists(path):  
            os.makedirs(path)  
  
        # self.seq = seq  
        self.path = path  
        self.header = {"Accept": "application/octet-stream"}  
        # 替换为你的服务器 URL  
        self.url = 'https://api.esmatlas.com/foldSequence/v1/pdb/'  
  
    def send_seq(self, data, prefix):  
        try:  
            res = requests.post(url=self.url, headers=self.header, data=data, verify=False)  
            res.raise_for_status()  # 如果响应状态码不是 200，将引发 HTTPError 异常  
            file_path = os.path.join(self.path, f'{prefix}.pdb')  
            with open(file_path, 'w') as file:  
                file.write(res.text)  
            return file_path  
        except requests.RequestException as e:  
            print(f"Error sending sequence: {e}")  
            return None  
```

```python
# 加载 CSV 文件  
root_path = 'H16'
result_path = root_path + '/PDB'
txt_path = root_path +'/'+ root_path + '.txt'
# df = pd.read_txt(root_path +'/'+ root_path + '.txt')  
  
# 筛选出长度大于 13 的序列  
# print(len(df))
# 初始化 esmfold_request 对象  
# path_to_save = 'output_pdbs'  
esmfold = esmfold_request(result_path)  

  
# 打开文件并读取  
with open(txt_path, 'r') as file:  
    # 迭代文件中的每一行  
    for index, line in tqdm(enumerate(file)):  
        # 去除每行末尾的换行符或回车符  
        sequence = line.strip()  
        # 打印肽链序列或进行其他处理  
        # print(peptide_sequence)
        prefix = f"seq_{index}"  
        # 发送序列并尝试保存结果  
        file_path = esmfold.send_seq(sequence, prefix)  

        time.sleep(2) 
  
```
